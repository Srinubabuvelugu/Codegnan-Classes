```python
import jwt
from datetime import datetime, timedelta, timezone

SECRET_KEY = "my-secret-key"

payload = {
    "user_id": 123,
    "email": "user@example.com",
    "role": "admin",
    "exp": datetime.now() + timedelta(hours=1)
}

token = jwt.encode(payload, SECRET_KEY, algorithm="HS256")

print(token)
```

```python
def outer(fun):
    def inner():
        print("Before")
        print(fun())
        print("After")
    return inner
@outer
def fun():
    return "Srinu"

print(fun())
```

```python
def outer(fun):
    def inner(x):
        if x!=0:
            return fun(x, x)
        else:
            return "Zero division not possiable"
    return inner
```

```python

@outer
def div(a, b):
    return a/b

print(div(5, 5))
print(div(5,0))
```

```python
from flask import Flask, session, redirect
app = Flask(__name__)

session['id'] = 2
@app.route("/")
def home():
    return "This is home page"
@app.route('/login')
def login():
    if "id" not in session:
        return redirect("/")
    return "This is Login page"
    
if __name__ == "__main__":
    app.run(debug=True)
```
